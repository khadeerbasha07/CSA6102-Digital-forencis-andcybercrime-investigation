```
#3. Write a Python program to detect phishing emails based on suspicious indicators in the sender, subject, and body.

import re


def check_email(sender, subject, body):
    flags = []


    if re.search(r"(urgent|verify your account|suspended|click here)", body, re.IGNORECASE):
        flags.append("Urgency/pressure language detected")


    if re.search(r"\d{1,3}(?:\.\d{1,3}){3}", body):
        flags.append("Raw IP address link found in body")


    domain = sender.split("@")[-1]
    if any(keyword in domain.lower() for keyword in ["secure", "verify", "update"]) and "@gmail" not in sender.lower():
        flags.append("Suspicious sender domain naming pattern")

    return flags



emails = [
    (
        "support@paypal.com",
        "Your monthly statement",
        "Please find your statement attached."
    ),
    (
        "alert@paypal-secure-verify.com",
        "URGENT: Verify your account",
        "Click here: http://192.168.10.5/login"
    ),
]

for sender, subject, body in emails:
    issues = check_email(sender, subject, body)
    verdict = "PHISHING SUSPECTED" if issues else "Looks legitimate"

    print(f"\nFrom: {sender}")
    print(f"Subject: {subject}")
    print(f"Verdict: {verdict}")

    for issue in issues:
        print(" -", issue)
```

```
#14.Digital Evidence Collection & Cataloguing

import os
import hashlib

# Create evidence directory
evidence_dir = "digital_evidence"
os.makedirs(evidence_dir, exist_ok=True)

# Sample evidence files
sample_evidence = {
    "email_log.txt": """From: attacker@mail.com
To: victim@mail.com
Subject: Invoice
Please pay immediately.""",

    "browser_history.txt": """http://malicious-site.com/login
http://bank.com/transfer""",

    "chat_message.txt": "Hey, did you send the file?",

    "system_log.txt": "2026-07-24 10:15:32 - USB device connected: E:\\"
}

# Create the evidence files
for filename, content in sample_evidence.items():
    with open(os.path.join(evidence_dir, filename), "w") as f:
        f.write(content)

# Function to catalog evidence
def catalog_evidence(folder):
    catalog = []

    for fname in sorted(os.listdir(folder)):
        path = os.path.join(folder, fname)

        stat = os.stat(path)

        with open(path, "rb") as f:
            data = f.read()

        catalog.append({
            "filename": fname,
            "size_bytes": stat.st_size,
            "md5": hashlib.md5(data).hexdigest()
        })

    return catalog

# Generate and display evidence catalog
evidence_catalog = catalog_evidence(evidence_dir)

for item in evidence_catalog:
    print(item)
```

```
#15.Volatile vs Non-Volatile Evidence Capture

import os
import psutil

# Function to capture volatile evidence (running processes)
def capture_volatile_evidence():
    processes = []

    for p in psutil.process_iter(['pid', 'name']):
        try:
            processes.append(p.info)
        except (psutil.NoSuchProcess, psutil.AccessDenied, psutil.ZombieProcess):
            pass

    return {
        "running_process_count": len(processes),
        "sample_processes": processes[:5]
    }

# Function to capture non-volatile evidence (files on disk)
def capture_nonvolatile_evidence(folder):
    files = sorted(os.listdir(folder))
    return {
        "disk_files": files,
        "file_count": len(files)
    }

# Folder created in the previous program
evidence_dir = "digital_evidence"

# Capture evidence
volatile_snapshot = capture_volatile_evidence()
nonvolatile_snapshot = capture_nonvolatile_evidence(evidence_dir)

# Display results
print("Volatile evidence -> running processes:",
      volatile_snapshot["running_process_count"])

print("Sample running processes:")
for process in volatile_snapshot["sample_processes"]:
    print(process)

print("\nNon-volatile evidence:")
print(nonvolatile_snapshot)
```

```
#16.Physical, Logical and Sparse Acquisition

import os

# Create source directory
os.makedirs("acquisition_demo/source", exist_ok=True)
source = "acquisition_demo/source"

# Create sample files
with open(os.path.join(source, "report.docx"), "w") as f:
    f.write("Confidential quarterly report.")

with open(os.path.join(source, "photo.jpg"), "w") as f:
    f.write("FAKEJPEGDATA")

# Simulated raw disk image (includes deleted data)
raw_disk = (
    b"REPORTDOCX_CONTENT"
    + b"\x00" * 20
    + b"DELETED_INVOICE_DATA"
    + b"\x00" * 10
)

# Physical acquisition: capture every byte
def physical_acquisition(raw_bytes):
    return bytes(raw_bytes)

# Logical acquisition: capture existing files
def logical_acquisition(folder):
    data = {}
    for filename in os.listdir(folder):
        path = os.path.join(folder, filename)
        with open(path, "rb") as f:
            data[filename] = f.read()
    return data

# Sparse acquisition: capture only selected files
def sparse_acquisition(folder, targets):
    data = {}
    for filename in targets:
        path = os.path.join(folder, filename)
        if os.path.exists(path):
            with open(path, "rb") as f:
                data[filename] = f.read()
    return data

# Perform acquisitions
physical_image = physical_acquisition(raw_disk)
logical_image = logical_acquisition(source)
sparse_image = sparse_acquisition(source, ["report.docx"])

# Display results
print("Physical image size:", len(physical_image), "bytes")
print("Logical image files:", list(logical_image.keys()))
print("Sparse image files:", list(sparse_image.keys()))
```

```
#17.Live vs Dead Acquisition

import time
import psutil
import json


def live_acquisition():
    return {
        "timestamp": time.time(),
        "running_processes": len(psutil.pids()),
        "cpu_percent": psutil.cpu_percent(interval=0.1),
    }


def dead_acquisition(snapshot_file):
    with open(snapshot_file, "r") as f:
        return json.load(f)


# Create a static snapshot file
dead_snapshot_path = "system_snapshot.json"

static_snapshot = {
    "hard_disk_files": ["a.txt", "b.txt"],
    "ram_data": None
}

with open(dead_snapshot_path, "w") as f:
    json.dump(static_snapshot, f, indent=4)


# Perform live and dead acquisition
live_result = live_acquisition()
dead_result = dead_acquisition(dead_snapshot_path)

# Display results
print("Live acquisition:")
print(live_result)

print("\nDead acquisition:")
print(dead_result)
```

```
#19.Hashing & the Avalanche Effect

import os
import hashlib
original_path = "original_evidence.bin"

with open(original_path, "wb") as f:
    f.write(os.urandom(1024))

def bit_stream_copy(src_path, dst_path):
    """Creates a bit-by-bit copy of the source file."""
    with open(src_path, "rb") as src, open(dst_path, "wb") as dst:
        dst.write(src.read())
copy_path = "bitstream_copy.bin"
bit_stream_copy(original_path, copy_path)


def sha256_of_file(path):
    """Calculates the SHA-256 hash of a file."""
    with open(path, "rb") as f:
        return hashlib.sha256(f.read()).hexdigest()

original_hash = sha256_of_file(original_path)
copy_hash = sha256_of_file(copy_path)

print("Original hash:", original_hash)
print("Copy hash:    ", copy_hash)

if original_hash == copy_hash:
    print("\nVerification Successful: The bit-stream copy is identical to the original.")
else:
    print("\nVerification Failed: The copy differs from the original.")
```

```
#20.File Carving (Data Recovery)
import os


def make_fake_jpeg(payload: bytes):
    """Create a fake JPEG with valid start and end markers."""
    return b"\xff\xd8\xff" + payload + b"\xff\xd9"


# Create two fake JPEG files
jpeg1 = make_fake_jpeg(b"PHOTO_OF_SUSPECT_CAR")
jpeg2 = make_fake_jpeg(b"CCTV_FRAME_CAPTURE")

# Simulate a raw disk containing random data and embedded JPEGs
raw_disk_blob = (
    os.urandom(30)
    + jpeg1
    + os.urandom(40)
    + jpeg2
    + os.urandom(20)
)


def carve_jpegs(blob: bytes):
    """Recover JPEG files from a raw binary blob."""
    recovered = []
    start_marker = b"\xff\xd8\xff"
    end_marker = b"\xff\xd9"

    pos = 0

    while True:
        start = blob.find(start_marker, pos)
        if start == -1:
            break

        end = blob.find(end_marker, start)
        if end == -1:
            break

        end += len(end_marker)
        recovered.append(blob[start:end])
        pos = end

    return recovered


# Recover JPEG files
recovered_files = carve_jpegs(raw_disk_blob)

# Display recovered files
print("Recovered JPEG files:", len(recovered_files))

for i, recovered in enumerate(recovered_files, start=1):
    print(f"\nRecovered JPEG {i}:")
    print(recovered)
```

```
#21.File System Feature Simulation (FAT vs NTFS vs EXT)
class FAT:
    def __init__(self):
        self.files = {}

    def create_file(self, name, content):
        self.files[name] = content
        print(f"[FAT] File '{name}' created.")

    def list_files(self):
        print("\n[FAT] Files:")
        for name in self.files:
            print("-", name)

    def features(self):
        print("\nFAT Features:")
        print("- Maximum simplicity")
        print("- No file permissions")
        print("- No journaling")
        print("- Compatible with most operating systems")


class NTFS:
    def __init__(self):
        self.files = {}

    def create_file(self, name, content, permission="Read/Write"):
        self.files[name] = {
            "content": content,
            "permission": permission
        }
        print(f"[NTFS] File '{name}' created.")

    def list_files(self):
        print("\n[NTFS] Files:")
        for name, info in self.files.items():
            print(f"- {name} (Permission: {info['permission']})")

    def features(self):
        print("\nNTFS Features:")
        print("- File permissions")
        print("- Journaling support")
        print("- Encryption")
        print("- Compression")
        print("- Large file support")


class EXT:
    def __init__(self):
        self.files = {}

    def create_file(self, name, content, owner="root"):
        self.files[name] = {
            "content": content,
            "owner": owner
        }
        print(f"[EXT] File '{name}' created.")

    def list_files(self):
        print("\n[EXT] Files:")
        for name, info in self.files.items():
            print(f"- {name} (Owner: {info['owner']})")

    def features(self):
        print("\nEXT Features:")
        print("- Linux file permissions")
        print("- Journaling")
        print("- Symbolic links")
        print("- High performance")
        print("- Large storage support")


# -------------------------------
# Demonstration
# -------------------------------

fat = FAT()
fat.create_file("report.txt", "Annual Report")
fat.list_files()
fat.features()

print("\n" + "=" * 50)

ntfs = NTFS()
ntfs.create_file("evidence.docx", "Forensic Report", "Read Only")
ntfs.list_files()
ntfs.features()

print("\n" + "=" * 50)

ext = EXT()
ext.create_file("system.log", "Linux Log File", "admin")
ext.list_files()
ext.features()
```

```
#22.Slack Space Calculation
CLUSTER_SIZE = 4096  # Typical disk cluster size in bytes


def calculate_slack_space(file_size, cluster_size=CLUSTER_SIZE):

    clusters_needed = -(-file_size // cluster_size)

    allocated_space = clusters_needed * cluster_size
    slack_space = allocated_space - file_size

    return allocated_space, slack_space

file_size = 5000  # bytes

allocated, slack = calculate_slack_space(file_size)

print(f"File size: {file_size} bytes")
print(f"Allocated space: {allocated} bytes")
print(f"Slack space: {slack} bytes")
```

```
#23.End-to-End Forensic Workflow
import hashlib
import json


def acquire(path):
    """Acquire evidence from a file."""
    with open(path, "rb") as f:
        return f.read()


def hash_data(data):
    """Generate SHA-256 hash of the data."""
    return hashlib.sha256(data).hexdigest()


def verify_integrity(data, expected_hash):
    """Verify data integrity using SHA-256."""
    return hash_data(data) == expected_hash


def analyze(data, keyword: bytes):
    """Check whether a keyword exists in the evidence."""
    return keyword in data


def generate_report(case_id, file_hash, keyword_found):
    """Generate a forensic analysis report."""
    return {
        "case_id": case_id,
        "sha256": file_hash,
        "suspicious_keyword_found": keyword_found,
        "status": "Evidence Verified" if keyword_found else "No Match",
    }


evidence_path = "workflow_evidence.txt"

with open(evidence_path, "wb") as f:
    f.write(b"Transfer $50000 to account 99881122 immediately, do not report.")

acquired_data = acquire(evidence_path)

evidence_hash = hash_data(acquired_data)

integrity_ok = verify_integrity(acquired_data, evidence_hash)

keyword_found = analyze(acquired_data, b"99881122")

report = generate_report(
    "CASE-2026-014",
    evidence_hash,
    keyword_found
)

print("Integrity Verified:", integrity_ok)

print(json.dumps(report, indent=4))
```

```
#24.Phishing Email Analysis
import re
import json
from urllib.parse import urlparse

# -------------------------------
# Sample Phishing Email
# -------------------------------

email_data = {
    "From": "PayPal Security <security@paypa1.com>",
    "Reply-To": "support@secure-login-paypal.com",
    "Subject": "Urgent: Verify Your Account",
    "Body": """
Dear Customer,

Your PayPal account has been temporarily suspended.

Verify immediately:
http://paypa1-login-security.com/verify

Alternative login:
https://secure-paypal-login.xyz/login

Official website:
https://www.paypal.com

Failure to verify within 24 hours will result in account suspension.

Regards,
PayPal Security Team
"""
}

# -------------------------------
# Extract URLs
# -------------------------------

def extract_urls(text):
    pattern = r'https?://[^\s]+'
    return re.findall(pattern, text)


# -------------------------------
# Extract Domain
# -------------------------------

def get_domain(url):
    return urlparse(url).netloc.lower()


# -------------------------------
# URL Classification
# -------------------------------

def classify_url(domain):
    suspicious_keywords = [
        "login",
        "secure",
        "verify",
        "update",
        "account",
        "bank"
    ]

    suspicious_tlds = [
        ".xyz",
        ".top",
        ".ru",
        ".tk",
        ".ml"
    ]

    reasons = []

    for keyword in suspicious_keywords:
        if keyword in domain:
            reasons.append(f"Contains keyword '{keyword}'")

    for tld in suspicious_tlds:
        if domain.endswith(tld):
            reasons.append(f"Uses suspicious TLD '{tld}'")

    if "paypa1" in domain:
        reasons.append("Possible typo-squatting detected")

    return reasons


# -------------------------------
# Sender Verification
# -------------------------------

def sender_spoof_check(sender):
    sender = sender.lower()

    if "paypa1" in sender:
        return True

    if "paypal.com" not in sender:
        return True

    return False


# -------------------------------
# Risk Score Calculation
# -------------------------------

def calculate_score(spoofed, url_results):

    score = 0

    if spoofed:
        score += 40

    for item in url_results:
        if item["Suspicious"]:
            score += 20

    return min(score, 100)


# -------------------------------
# Analysis
# -------------------------------

urls = extract_urls(email_data["Body"])

url_results = []

for url in urls:

    domain = get_domain(url)

    reasons = classify_url(domain)

    url_results.append({
        "URL": url,
        "Domain": domain,
        "Suspicious": len(reasons) > 0,
        "Reasons": reasons
    })

spoofed = sender_spoof_check(email_data["From"])

risk_score = calculate_score(spoofed, url_results)

# -------------------------------
# Report
# -------------------------------

report = {
    "Sender": email_data["From"],
    "Reply-To": email_data["Reply-To"],
    "Subject": email_data["Subject"],
    "Spoofed Sender": spoofed,
    "Risk Score": risk_score,
    "Indicators of Compromise": url_results,
    "Summary": {
        "Total URLs": len(url_results),
        "Suspicious URLs": sum(
            item["Suspicious"] for item in url_results
        ),
        "Possible Typosquatting": any(
            "typo" in " ".join(item["Reasons"]).lower()
            for item in url_results
        )
    }
}

print("========== PHISHING EMAIL REPORT ==========")
print(json.dumps(report, indent=4))
```

```
#25.Malware Hash Analysis
import hashlib
import csv
import os

# ----------------------------------------
# Sample Files
# ----------------------------------------

samples = {
    "file1.exe": b"Windows executable content",
    "file2.dll": b"Dynamic link library",
    "malware1.exe": b"EICAR_TEST_FILE",
    "malware2.exe": b"RANSOMWARE_SAMPLE"
}

# ----------------------------------------
# Hash Functions
# ----------------------------------------

def md5_hash(data):
    return hashlib.md5(data).hexdigest()


def sha1_hash(data):
    return hashlib.sha1(data).hexdigest()


def sha256_hash(data):
    return hashlib.sha256(data).hexdigest()


# ----------------------------------------
# Create Sample Files
# ----------------------------------------

for filename, content in samples.items():
    with open(filename, "wb") as f:
        f.write(content)

# ----------------------------------------
# Malware Database Creation
# ----------------------------------------

database = []

for filename in samples:

    with open(filename, "rb") as f:
        data = f.read()

    md5 = md5_hash(data)
    sha1 = sha1_hash(data)
    sha256 = sha256_hash(data)

    malware = filename.startswith("malware")

    database.append([
        filename,
        md5,
        sha1,
        sha256,
        malware
    ])

# ----------------------------------------
# Save Database
# ----------------------------------------

csv_file = "malware_hash_database.csv"

with open(csv_file, "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow([
        "Filename",
        "MD5",
        "SHA1",
        "SHA256",
        "Known_Malware"
    ])

    writer.writerows(database)

print("Malware hash database saved as:", csv_file)

# ----------------------------------------
# Detection System
# ----------------------------------------

known_hashes = {
    row[3] for row in database if row[4]
}

detected = 0

print("\nDetection Results")

for row in database:

    filename = row[0]
    sha256 = row[3]

    if sha256 in known_hashes:
        print(filename, "-> Malware Detected")
        detected += 1
    else:
        print(filename, "-> Clean")

# ----------------------------------------
# Statistics
# ----------------------------------------

total = len(database)

print("\nStatistics")
print("Total Files:", total)
print("Detected Malware:", detected)
print("Detection Rate: {:.2f}%".format((detected / total) * 100))

# ----------------------------------------
# Hash Family Clustering
# ----------------------------------------

print("\nHash Family Clustering")
print("MD5 Hashes:", total)
print("SHA-1 Hashes:", total)
print("SHA-256 Hashes:", total)

# ----------------------------------------
# Display Database
# ----------------------------------------

print("\nDatabase Entries")

for row in database:
    print("------------------------------")
    print("Filename :", row[0])
    print("MD5      :", row[1])
    print("SHA1     :", row[2])
    print("SHA256   :", row[3])
    print("Malware  :", row[4])

# ----------------------------------------
# Cleanup (Optional)
# ----------------------------------------

# Uncomment to remove sample files
#
# for filename in samples:
#     os.remove(filename)
```

```
#27.Hacking Attack Simulation (Ethical)
import re
import json
from collections import Counter

# ----------------------------------------
# Sample Login Attempts (Simulation)
# ----------------------------------------

login_attempts = [
    ("alice", "123456"),
    ("alice", "password"),
    ("alice", "admin"),
    ("alice", "welcome"),
    ("alice", "CorrectPassword123")
]

# ----------------------------------------
# Sample Web Requests
# ----------------------------------------

requests = [
    "/login?user=admin' OR '1'='1",
    "/products?id=10",
    "/../../etc/passwd",
    "/images/logo.png",
    "/download?file=../../../windows/system32/config",
    "/search?q=laptop"
]

# ----------------------------------------
# Weak Password List
# ----------------------------------------

weak_passwords = {
    "123456",
    "password",
    "admin",
    "welcome",
    "qwerty",
    "abc123"
}

# ----------------------------------------
# Brute Force Simulation
# ----------------------------------------

attack_logs = []

failed_attempts = 0
RATE_LIMIT = 4

for username, password in login_attempts:

    if password != "CorrectPassword123":
        failed_attempts += 1

        attack_logs.append({
            "Attack": "Brute Force",
            "User": username,
            "Password": password,
            "Status": "Failed"
        })

    else:
        attack_logs.append({
            "Attack": "Login",
            "User": username,
            "Password": "********",
            "Status": "Success"
        })

if failed_attempts >= RATE_LIMIT:
    attack_logs.append({
        "Attack": "Rate Limit",
        "Status": "Too many failed attempts"
    })

# ----------------------------------------
# Weak Password Analysis
# ----------------------------------------

weak_found = []

for username, password in login_attempts:
    if password in weak_passwords:
        weak_found.append(password)

# ----------------------------------------
# SQL Injection Detection
# ----------------------------------------

sql_pattern = re.compile(
    r"(union\s+select|or\s+'?1'='?1|--|drop\s+table)",
    re.IGNORECASE
)

# ----------------------------------------
# Directory Traversal Detection
# ----------------------------------------

traversal_pattern = re.compile(r"\.\./")

# ----------------------------------------
# Analyze Requests
# ----------------------------------------

findings = []

for req in requests:

    if sql_pattern.search(req):
        findings.append({
            "Type": "SQL Injection",
            "Request": req,
            "Severity": "High"
        })

    if traversal_pattern.search(req):
        findings.append({
            "Type": "Directory Traversal",
            "Request": req,
            "Severity": "High"
        })

# ----------------------------------------
# Vulnerability Report
# ----------------------------------------

report = {
    "Weak Passwords": len(weak_found),
    "Failed Login Attempts": failed_attempts,
    "SQL Injection Attempts":
        sum(f["Type"] == "SQL Injection" for f in findings),
    "Directory Traversal Attempts":
        sum(f["Type"] == "Directory Traversal" for f in findings)
}

# ----------------------------------------
# WAF Suggestions
# ----------------------------------------

waf_rules = [
    "Block SQL keywords such as UNION SELECT and DROP TABLE.",
    "Block '../' directory traversal sequences.",
    "Enable login rate limiting after multiple failures.",
    "Require strong password policy.",
    "Enable Web Application Firewall logging.",
    "Use parameterized SQL queries."
]

# ----------------------------------------
# Output
# ----------------------------------------

print("=" * 60)
print("ATTACK PATTERN LOGS")
print("=" * 60)

print(json.dumps(attack_logs, indent=4))

print("\n" + "=" * 60)
print("VULNERABILITY CLASSIFICATION REPORT")
print("=" * 60)

print(json.dumps(report, indent=4))

print("\n" + "=" * 60)
print("REQUEST FINDINGS")
print("=" * 60)

print(json.dumps(findings, indent=4))

print("\n" + "=" * 60)
print("WAF RULE SUGGESTIONS")
print("=" * 60)

for i, rule in enumerate(waf_rules, start=1):
    print(f"{i}. {rule}")
```
